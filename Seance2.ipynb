```python
"""
@author: Nuno Prata - [M2 IDIM] 2020-2021

The code calculates,saves as csv, loads and displays the optimized input VE for an RC filter
as well as the output VC. The voltage of the resistance R is obtained by VR=VE-VC.

E0       : desired input voltage
fexc     : frequency of the excitation
R        : value of the resistance of the filter
C        : value of the capacitance of the filter
Tacq     : duration of the acquisition
TE       : desired stabilisation delay
Fs       : frequency of sampling
filename : name of the storing file
"""
# Importer les sous-programmes nécessaires:
from CREATE_DataOpt import createOpt
from SAVE_Data import save
from LOAD_Data import load
from DISPLAY_Data import display

# Initialisation des entrées
E0 = 5
fexc = 20000; Texc = 1 / fexc  # Texc is the period of the excitation (optionnal parameter)
R = 10000
C = 45e-9
Tacq = 20*Texc
TE = Texc/4;
Fs = 1e6
filename = 'Shortcut_Py.csv'

# Procédure de résolution
t, VE, VC = createOpt(E0, fexc, R, C, Tacq, Fs, TE)
save(filename, t, VE, VC)
t1, VE1, VC1 = load(filename)
display(t1, VE1, VC1)
```
